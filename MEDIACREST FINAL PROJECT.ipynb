## ***TASK 3: MODEL EVALUATION  AND COMPARISON*** ##

### **(a) Performance Metrics** ##

### **(i) Importing the Evaluation Metrics** ###

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

import pandas as pd
```

### **(ii) Creating a Function to Evaluate Any Model** ###

```python
def evaluate_model(y_true, y_pred, y_prob):

    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    roc_auc = roc_auc_score(y_true, y_prob)

    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

    specificity = tn / (tn + fp)

    return [
        accuracy,
        precision,
        recall,
        f1,
        roc_auc,
        specificity
    ]
```

### **(iii) Evaluating Every Model** ###

```python
results = pd.DataFrame({

    "Logistic Regression":
        evaluate_model(y_test, y_pred_log, y_prob_log),

    "KNN":
        evaluate_model(y_test, y_pred_knn, y_prob_knn),

    "SVM":
        evaluate_model(y_test, y_pred_svm, y_prob_svm),

    "Random Forest":
        evaluate_model(y_test, y_pred_rf, y_prob_rf)

},

index=[
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC",
    "Specificity"
])

results
```

```python
results = results.round(3)

results
```

```python
        ##  Accuracy
# Accuracy measures the overall proportion of correctly classified patients.
#It provides a general indication of model performance but may be misleading when class imbalance exists.
      ##Precision
#Precision measures the proportion of patients predicted to have hypertension who truly have hypertension.
#High precision reduces false positive diagnoses.
    ##Recall (Sensitivity)
#Recall measures the proportion of actual hypertension cases correctly identified by the model.
# High recall is especially important in healthcare because missing patients with hypertension could delay treatment.
    ##F1-Score
#The F1-score balances precision and recall, making it suitable for evaluating models on datasets with class imbalance.
   ##ROC-AUC
#ROC-AUC measures the model's ability to distinguish between patients with and without hypertension across different classification thresholds
## Higher values indicate better discrimination.
  ##Specificity
#Specificity measures the proportion of patients without hypertension who are correctly identified.
#High specificity reduces unnecessary follow-up investigations for healthy individuals.
```

### **(b) Cross-Validation and ROC Curves** ##

### **(i) Importing the Libraries** ###

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_val_score

import matplotlib.pyplot as plt
```

### **(ii) Creating Stratified K-Fold** ###

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

### **(iii) Cross-Validation Scores** ###

```python
##FOR LOGISTIC REGRESSION
log_scores = cross_val_score(
    log_model,
    X_train_scaled,
    y_train,
    cv=cv,
    scoring='roc_auc'
)
```

```python
##FOR KNN
knn_scores = cross_val_score(
    knn_model,
    X_train_scaled,
    y_train,
    cv=cv,
    scoring='roc_auc'
)
```

```python
##FOR SVM
svm_scores = cross_val_score(
    svm_model,
    X_train_scaled,
    y_train,
    cv=cv,
    scoring='roc_auc'
)
```

```python
#FOR RANDOM FOREST
rf_scores = cross_val_score(
    rf_model,
    X_train,
    y_train,
    cv=cv,
    scoring='roc_auc'
)
```

### **(iV) Displaying results** ###

```python
cv_results = pd.DataFrame({

    "Mean ROC-AUC":[
        log_scores.mean(),
        knn_scores.mean(),
        svm_scores.mean(),
        rf_scores.mean()
    ],

    "Std Dev":[
        log_scores.std(),
        knn_scores.std(),
        svm_scores.std(),
        rf_scores.std()
    ]

},

index=[
    "Logistic Regression",
    "KNN",
    "SVM",
    "Random Forest"
])

cv_results.round(3)
```
